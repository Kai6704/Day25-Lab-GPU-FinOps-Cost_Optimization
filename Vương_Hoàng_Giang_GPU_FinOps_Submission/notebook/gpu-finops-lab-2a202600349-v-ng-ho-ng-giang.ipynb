## Part 7: Complete FinOps Workflow
Run a full cycle: submit workloads → monitor → detect waste → autoscale → optimize cost.

```python
import requests
import json

# Hàm bổ trợ để xử lý lỗi "Extra data" từ Cloudflare Tunnel
def get_clean_json(response):
    try:
        # Thử lấy dòng đầu tiên nếu có nhiều dòng (đặc thù lỗi Tunnel)
        first_line = response.text.strip().split('\n')[0]
        return json.loads(first_line)
    except Exception:
        # Nếu vẫn lỗi, thử parse toàn bộ (cách chuẩn)
        return response.json()

print("🔄 FULL FINOPS OPTIMIZATION WORKFLOW")
print("=" * 60)

# Step 1: Check initial state
print("\n1️⃣  Initial cluster state:")
r = requests.get(f"{GATEWAY_URL}/cluster/metrics")
m = get_clean_json(r)
print(f"   GPUs: {m['total_gpus']} | Util: {m['avg_utilization']:.1f}% | Idle: {m['idle_gpus']}")

# Step 2: Submit heavy workloads
print("\n2️⃣  Submitting heavy workloads...")
for i in range(5):
    requests.post(f"{GATEWAY_URL}/cluster/workloads/submit", json={
        "workload_id": f"heavy-{i:03d}", "gpu_type_preferred": "T4", "gpu_count": 1, "duration_seconds": 600
    })
r = requests.get(f"{GATEWAY_URL}/cluster/metrics")
m = get_clean_json(r)
print(f"   After load: Util: {m['avg_utilization']:.1f}% | Busy: {m['busy_gpus']}/{m['total_gpus']}")

# Step 3: Autoscaler evaluates
print("\n3️⃣  Autoscaler evaluation:")
r = requests.post(f"{GATEWAY_URL}/autoscaler/evaluate")
decision = get_clean_json(r)
print(f"   Decision: {decision['action']} - {decision['reason']}")

# Step 4: Cost snapshot
print("\n4️⃣  Cost analysis:")
r = requests.post(f"{GATEWAY_URL}/cost/snapshot")
snap = get_clean_json(r)
print(f"   Total cost/interval: ${snap['total_cost_usd']:.6f}")
print(f"   Waste: {snap['waste_pct']:.1f}%")

# Step 5: Get recommendations
print("\n5️⃣  Recommendations:")
r = requests.post(f"{GATEWAY_URL}/cost/recommendations")
recs = get_clean_json(r)
for rec in recs:
    print(f"   [{rec['priority']}] {rec['type']}: savings ~{rec['estimated_savings_pct']}%")

# Step 6: Apply optimization
print("\n6️⃣  Applying optimization: Switch to spot instances...")
for i in range(3):
    requests.post(f"{GATEWAY_URL}/spot/request", json={
        "instance_id": f"opt-spot-{i}", "gpu_type": "T4",
        "gpu_count": 1, "max_price_per_hour": 0.15, "workload_id": f"heavy-{i:03d}"
    })
r = requests.get(f"{GATEWAY_URL}/spot/savings-report")
savings = get_clean_json(r)
print(f"   Spot savings: ${savings['total_savings']:.4f} ({savings['savings_pct']:.1f}%)")

# Step 7: Final billing
print("\n7️⃣  Final billing:")
for i in range(5):
    requests.post(f"{GATEWAY_URL}/cluster/workloads/heavy-{i:03d}/complete")
    requests.post(f"{GATEWAY_URL}/billing/record", json={
        "workload_id": f"heavy-{i:03d}", "gpu_type": "T4",
        "gpu_count": 1, "duration_seconds": 600, "is_spot": i < 3
    })

r = requests.get(f"{GATEWAY_URL}/billing/summary")
final = get_clean_json(r)

# Kiểm tra xem final có phải là dict không trước khi in
if isinstance(final, dict) and 'total_cost_usd' in final:
    print(f"   Total spend: ${final['total_cost_usd']:.4f}")
    print(f"   Total saved: ${final['total_savings_usd']:.4f}")
    print(f"   Budget: {final['budget_utilization_pct']:.1f}% used")
else:
    print(f"   ⚠️ Lỗi: Backend trả về dữ liệu không hợp lệ: {final}")

print(f"\n✅ Workflow complete!")
```

```python
# Cell 20: GPU Metrics Collection (works on Kaggle/Colab)
# Strategy: pynvml first (reliable Python lib), torch.cuda fallback

def get_gpu_metrics():
    """Get GPU metrics inline. Called during training loop."""

    # Method 1: pynvml (most reliable on Kaggle - pure Python, no subprocess)
    if PYNVML_AVAILABLE:
        try:
            nvmlInit()
            handle = nvmlDeviceGetHandleByIndex(0)
            util = nvmlDeviceGetUtilizationRates(handle)
            mem_info = nvmlDeviceGetMemoryInfo(handle)
            try:
                power = nvmlDeviceGetPowerUsage(handle) / 1000.0  # mW -> W
            except Exception:
                power = 0.0
            try:
                temp = nvmlDeviceGetTemperature(handle, 0)  # NVML_TEMPERATURE_GPU = 0
            except Exception:
                temp = 0.0
            return {
                'timestamp': time.time(),
                'gpu_util_pct': float(util.gpu),
                'mem_util_pct': float(util.memory),
                'mem_used_mb': mem_info.used / 1e6,
                'mem_total_mb': mem_info.total / 1e6,
                'power_watts': float(power),
                'temp_c': float(temp),
            }
        except Exception as e:
            pass  # Fall through to torch.cuda

    # Method 2: torch.cuda (always works, limited metrics)
    mem_used = torch.cuda.memory_allocated() / 1e6
    mem_reserved = torch.cuda.memory_reserved() / 1e6
    mem_total = torch.cuda.get_device_properties(0).total_memory / 1e6
    util_approx = (mem_reserved / mem_total) * 100 if mem_total > 0 else 0
    return {
        'timestamp': time.time(),
        'gpu_util_pct': util_approx,
        'mem_util_pct': util_approx,
        'mem_used_mb': mem_used,
        'mem_total_mb': mem_total,
        'power_watts': 0.0,
        'temp_c': 0.0,
    }

# === DIAGNOSTIC TEST ===
print("=" * 50)
print("GPU METRICS DIAGNOSTIC")
print("=" * 50)

# Test pynvml
print(f"\n1. pynvml available: {PYNVML_AVAILABLE}")
if PYNVML_AVAILABLE:
    try:
        nvmlInit()
        handle = nvmlDeviceGetHandleByIndex(0)
        util = nvmlDeviceGetUtilizationRates(handle)
        mem = nvmlDeviceGetMemoryInfo(handle)
        print(f"   pynvml works! GPU util={util.gpu}%, mem={mem.used/1e6:.0f}/{mem.total/1e6:.0f} MB")
        try:
            pwr = nvmlDeviceGetPowerUsage(handle) / 1000.0
            print(f"   Power: {pwr:.1f}W")
        except Exception as e:
            print(f"   Power: not available ({e})")
        try:
            tmp = nvmlDeviceGetTemperature(handle, 0)
            print(f"   Temp: {tmp}C")
        except Exception as e:
            print(f"   Temp: not available ({e})")
    except Exception as e:
        print(f"   pynvml FAILED: {e}")

# Test full function
print(f"\n2. get_gpu_metrics() test:")
test_result = get_gpu_metrics()
for k, v in sorted(test_result.items()):
    if k != 'timestamp':
        print(f"   {k}: {v}")

print(f"\n   Method: {'pynvml' if test_result.get('temp_c', 0) > 0 or (PYNVML_AVAILABLE and test_result.get('gpu_util_pct', -1) >= 0) else 'torch.cuda'}")
print(f"\nReady for training.")
```

```python
# Cell 21: Prepare CIFAR-10 dataset and ResNet-18 model
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
])

trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform_train)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2)

testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010)),
]))
testloader = torch.utils.data.DataLoader(testset, batch_size=256, shuffle=False, num_workers=2)

print(f"Dataset ready: {len(trainset)} train / {len(testset)} test images")
print(f"Batches per epoch: {len(trainloader)}")


def create_model():
    """Create a fresh ResNet-18 for CIFAR-10."""
    model = torchvision.models.resnet18(weights=None, num_classes=10)
    model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
    model.maxpool = nn.Identity()
    return model.cuda()


def train_epoch_monitored(model, loader, optimizer, criterion, use_amp=False, sample_interval=10):
    """Train one epoch with inline GPU metrics collection every N batches."""
    model.train()
    total_loss, correct, total = 0, 0, 0
    scaler = GradScaler() if use_amp else None
    metrics_samples = []
    start = time.time()

    for batch_idx, (inputs, targets) in enumerate(loader):
        inputs, targets = inputs.cuda(), targets.cuda()
        optimizer.zero_grad()

        if use_amp:
            with autocast():
                outputs = model(inputs)
                loss = criterion(outputs, targets)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()

        total_loss += loss.item() * inputs.size(0)
        _, predicted = outputs.max(1)
        correct += predicted.eq(targets).sum().item()
        total += inputs.size(0)

        # Collect GPU metrics every N batches (inline, no threading)
        if batch_idx % sample_interval == 0:
            torch.cuda.synchronize()
            sample = get_gpu_metrics()
            metrics_samples.append(sample)

    elapsed = time.time() - start
    return total_loss / total, correct / total * 100, elapsed, metrics_samples


print("Model & training functions ready (inline monitoring every 10 batches)")
```

```python
# Cell 22: Train FP32 (baseline) with inline GPU monitoring
NUM_EPOCHS = 3
price_per_hour = GPU_PRICING[detected_type]

print("=" * 60)
print("EXPERIMENT 1: FP32 Training (Baseline)")
print("=" * 60)

model_fp32 = create_model()
optimizer_fp32 = optim.SGD(model_fp32.parameters(), lr=0.1, momentum=0.9, weight_decay=5e-4)
criterion = nn.CrossEntropyLoss()

torch.cuda.synchronize()
torch.cuda.reset_peak_memory_stats()
fp32_start = time.time()

fp32_results = []
fp32_all_metrics = []
for epoch in range(NUM_EPOCHS):
    loss, acc, t, epoch_metrics = train_epoch_monitored(model_fp32, trainloader, optimizer_fp32, criterion, use_amp=False)
    fp32_results.append({"epoch": epoch+1, "loss": loss, "accuracy": acc, "time_s": t})
    fp32_all_metrics.extend(epoch_metrics)
    print(f"   Epoch {epoch+1}/{NUM_EPOCHS} | Loss: {loss:.4f} | Acc: {acc:.1f}% | Time: {t:.1f}s | Samples: {len(epoch_metrics)}")

fp32_total_time = time.time() - fp32_start
fp32_peak_mem = torch.cuda.max_memory_allocated() / 1e9
fp32_cost = (fp32_total_time / 3600) * price_per_hour

print(f"\n   Total samples collected: {len(fp32_all_metrics)}")

# Build monitor dataframe
fp32_monitor_df = pd.DataFrame(fp32_all_metrics)
if not fp32_monitor_df.empty and 'timestamp' in fp32_monitor_df.columns:
    fp32_monitor_df['elapsed'] = fp32_monitor_df['timestamp'] - fp32_monitor_df['timestamp'].iloc[0]

print(f"\n   FP32 Summary:")
print(f"   Total time:      {fp32_total_time:.1f}s")
print(f"   Peak memory:     {fp32_peak_mem:.2f} GB")
if not fp32_monitor_df.empty and 'gpu_util_pct' in fp32_monitor_df.columns:
    print(f"   Avg GPU util:    {fp32_monitor_df['gpu_util_pct'].mean():.1f}%")
    print(f"   Avg power:       {fp32_monitor_df['power_watts'].mean():.1f}W")
    print(f"   Avg temperature: {fp32_monitor_df['temp_c'].mean():.1f}C")
    print(f"   Max GPU util:    {fp32_monitor_df['gpu_util_pct'].max():.1f}%")
else:
    print(f"   GPU util:        DataFrame empty! columns={list(fp32_monitor_df.columns) if not fp32_monitor_df.empty else 'EMPTY'}")
    if fp32_all_metrics:
        print(f"   First sample keys: {list(fp32_all_metrics[0].keys())}")
print(f"   Estimated cost:  ${fp32_cost:.6f}")
```

```python
# Cell 27: Multi-GPU Cost Analysis
print("=" * 60)
print("EXERCISE 8.5.1: Multi-GPU Cost Analysis")
print("=" * 60)

# Giả định bảng giá GPU (Thường được định nghĩa ở các cell đầu của Lab)
GPU_PRICING = {
    "T4": 0.35,
    "A100": 2.48,
    "RTX3060": 0.15 # Giả định giá trị khấu hao cho máy trạm của bạn
}

def analyze_multi_gpu_cost(base_time_hours, gpu_type, gpu_counts, scaling_factors=None):
    price_per_hour = GPU_PRICING.get(gpu_type, 0.5)
    analysis = {}
    
    # Nếu không có scaling_factors, tạo một mô hình thực tế (hiệu suất giảm dần)
    # Ví dụ: 1 GPU=1.0x, 2 GPUs=1.85x, 4 GPUs=3.4x, 8 GPUs=6.0x
    if scaling_factors is None:
        scaling_factors = {1: 1.0, 2: 1.85, 4: 3.4, 8: 6.0}

    for count in gpu_counts:
        efficiency = scaling_factors.get(count, count * 0.8) # Mặc định hiệu suất 80% nếu thiếu
        
        # Thời gian huấn luyện thực tế = Thời gian gốc / Hệ số tăng tốc
        training_time = base_time_hours / efficiency
        
        # Tổng chi phí = Thời gian * Giá mỗi giờ * Số lượng GPU
        total_cost = training_time * price_per_hour * count
        
        # Chi phí trên mỗi đơn vị hiệu năng (Càng thấp càng tốt)
        # Ở đây ta lấy Speedup / Total Cost
        cost_perf_ratio = efficiency / total_cost if total_cost > 0 else 0
        
        analysis[count] = {
            "time_hours": training_time,
            "total_cost": total_cost,
            "efficiency_pct": (efficiency / count) * 100,
            "cost_perf_ratio": cost_perf_ratio
        }
    
    return analysis

# Test data
base_training_time = 24.0  # Giả sử training 1 ngày trên 1 GPU
test_gpu_counts = [1, 2, 4, 8]
target_gpu = "T4"

# Chạy phân tích
results = analyze_multi_gpu_cost(base_training_time, target_gpu, test_gpu_counts)

# Hiển thị kết quả dạng bảng
print(f"{'GPUs':<6} | {'Time (h)':<10} | {'Total Cost':<12} | {'Efficiency':<12} | {'Cost/Perf'}")
print("-" * 65)

optimal_count = 1
max_ratio = 0

for count in test_gpu_counts:
    data = results[count]
    print(f"{count:<6} | {data['time_hours']:<10.2f} | ${data['total_cost']:<11.2f} | {data['efficiency_pct']:<11.1f}% | {data['cost_perf_ratio']:.4f}")
    
    # Tìm cấu hình tối ưu nhất về chi phí
    if data['cost_perf_ratio'] > max_ratio:
        max_ratio = data['cost_perf_ratio']
        optimal_count = count

print("-" * 65)
print(f"💡 FINOPS INSIGHT: Optimal configuration for {target_gpu} is {optimal_count} GPU(s).")
print(f"   Reasoning: Adding more GPUs after this point increases cost faster than it reduces training time.")
```

```python
# Cell 28: Project Cost Forecasting
import math

print("=" * 60)
print("EXERCISE 8.5.2: Project Cost Forecasting")
print("=" * 60)

# Giả định bảng giá GPU (Đảm bảo đồng bộ với Cell trước)
GPU_PRICING = {
    "T4": 0.35,
    "A100": 2.48,
    "RTX3060": 0.15
}

def forecast_project_cost(phases, contingency_pct=20, confidence_level=0.95):
    forecast = {
        "phases_breakdown": [],
        "total_base_cost": 0.0,
        "total_uncertainty_variance": 0.0
    }
    
    for phase in phases:
        price = GPU_PRICING.get(phase['gpu_type'], 0.5)
        # Chi phí cơ sở của giai đoạn = thời gian * giá * số lượng
        base_cost = phase['duration_hours'] * price * phase['gpu_count']
        
        # Độ lệch chuẩn giả định dựa trên uncertainty_pct (Xấp xỉ 2 sigma cho mức độ không chắc chắn)
        # Giả định uncertainty_pct đại diện cho độ lệch chuẩn so với giá trị trung bình
        std_dev = base_cost * phase['uncertainty_pct']
        
        forecast["phases_breakdown"].append({
            "name": phase['name'],
            "base_cost": base_cost,
            "std_dev": std_dev
        })
        
        forecast["total_base_cost"] += base_cost
        # Tính tổng phương sai để xác định khoảng tin cậy cho tổng dự án
        forecast["total_uncertainty_variance"] += std_dev**2

    # Tính toán kết quả tổng hợp
    base_total = forecast["total_base_cost"]
    contingency = base_total * (contingency_pct / 100)
    expected_total = base_total + contingency
    
    # Tính Confidence Interval (Sử dụng Z-score cho 95% confidence ~ 1.96)
    z_score = 1.96 if confidence_level == 0.95 else 1.645 # Mặc định 95% hoặc 90%
    total_std_dev = math.sqrt(forecast["total_uncertainty_variance"])
    margin_of_error = z_score * total_std_dev
    
    return {
        "breakdown": forecast["phases_breakdown"],
        "base_cost": base_total,
        "contingency": contingency,
        "expected_total": expected_total,
        "best_case": max(0, expected_total - margin_of_error),
        "worst_case": expected_total + margin_of_error,
        "margin_of_error": margin_of_error
    }

# Example project phases
example_project = [
    {"name": "Data Preparation", "gpu_type": "T4", "gpu_count": 1, "duration_hours": 40, "uncertainty_pct": 0.15},
    {"name": "Model Training", "gpu_type": "A100", "gpu_count": 4, "duration_hours": 120, "uncertainty_pct": 0.25},
    {"name": "Hyperparameter Tuning", "gpu_type": "A100", "gpu_count": 8, "duration_hours": 60, "uncertainty_pct": 0.30},
    {"name": "Model Evaluation", "gpu_type": "T4", "gpu_count": 2, "duration_hours": 20, "uncertainty_pct": 0.10},
]

# Thực hiện dự báo
f = forecast_project_cost(example_project)

# Hiển thị kết quả
print(f"{'Phase Name':<25} | {'Base Cost':<12}")
print("-" * 40)
for p in f['breakdown']:
    print(f"{p['name']:<25} | ${p['base_cost']:>10.2f}")

print("-" * 40)
print(f"Subtotal (Base):          ${f['base_cost']:>10.2f}")
print(f"Contingency ({20}%):      ${f['contingency']:>10.2f}")
print(f"Expected Total:           ${f['expected_total']:>10.2f}")
print("-" * 40)
print(f"📊 95% CONFIDENCE INTERVAL:")
print(f"   Best Case:             ${f['best_case']:>10.2f}")
print(f"   Worst Case:            ${f['worst_case']:>10.2f}")
print(f"   Potential Variance:    ±${f['margin_of_error']:>10.2f}")
```

```python
# Cell 29: Optimization Opportunity Analysis
print("=" * 60)
print("EXERCISE 8.5.3: Advanced Optimization Opportunity Analysis")
print("=" * 60)

# Giả định bảng giá GPU
GPU_PRICING = {"T4": 0.35, "A100": 2.48, "RTX3060": 0.15}

def analyze_optimization_opportunities(current_config, optimization_strategies):
    # Quy đổi Effort và Risk thành điểm số để tính toán
    score_map = {"LOW": 1, "MEDIUM": 2, "HIGH": 3}
    
    # Tính chi phí hiện tại (Base Cost)
    base_price = GPU_PRICING.get(current_config['gpu_type'], 0.5)
    current_cost = current_config['duration_hours'] * base_price * current_config['gpu_count']
    
    results = []
    for strat in optimization_strategies:
        savings_amount = current_cost * strat['savings_pct']
        
        # Chỉ số ưu tiên: Savings_pct / Effort (Càng cao càng tốt)
        effort_val = score_map.get(strat['implementation_effort'], 2)
        priority_score = (strat['savings_pct'] * 100) / effort_val
        
        results.append({
            **strat,
            "potential_savings_usd": savings_amount,
            "priority_score": priority_score
        })
    
    # Sắp xếp theo điểm ưu tiên giảm dần
    ranked_strategies = sorted(results, key=lambda x: x['priority_score'], reverse=True)
    
    # Tính toán tích lũy (Cumulative Savings)
    # Lưu ý: Các khoản tiết kiệm không cộng dồn tuyến tính (ví dụ: giảm 50% rồi giảm tiếp 20% là 0.5 * 0.8)
    cumulative_multiplier = 1.0
    roadmap = []
    
    for strat in ranked_strategies:
        cumulative_multiplier *= (1 - strat['savings_pct'])
        cumulative_savings_pct = (1 - cumulative_multiplier) * 100
        
        roadmap.append({
            "name": strat['name'],
            "effort": strat['implementation_effort'],
            "risk": strat['risk_level'],
            "individual_savings_pct": strat['savings_pct'] * 100,
            "cumulative_savings_pct": cumulative_savings_pct
        })
        
    return {
        "current_cost": current_cost,
        "ranked_strategies": ranked_strategies,
        "roadmap": roadmap,
        "total_potential_savings_pct": (1 - cumulative_multiplier) * 100
    }

# Data mẫu
example_strategies = [
    {"name": "Switch to Mixed Precision (AMP)", "savings_pct": 0.25, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Use Spot Instances", "savings_pct": 0.60, "implementation_effort": "MEDIUM", "risk_level": "HIGH", "dependencies": []},
    {"name": "Optimize Batch Size", "savings_pct": 0.15, "implementation_effort": "LOW", "risk_level": "LOW", "dependencies": []},
    {"name": "Implement Early Stopping", "savings_pct": 0.20, "implementation_effort": "MEDIUM", "risk_level": "LOW", "dependencies": []},
    {"name": "Switch to More Efficient GPU", "savings_pct": 0.40, "implementation_effort": "HIGH", "risk_level": "MEDIUM", "dependencies": []},
]

current_training_config = {
    "gpu_type": "A100", "gpu_count": 4, "duration_hours": 100,
    "precision": "FP32", "instance_type": "on-demand",
}

# Thực hiện phân tích
analysis = analyze_optimization_opportunities(current_training_config, example_strategies)

print(f"💰 Current Estimated Cost: ${analysis['current_cost']:.2f}")
print(f"🚀 Maximum Cumulative Savings: {analysis['total_potential_savings_pct']:.1f}%\n")

print(f"{'Order':<5} | {'Strategy Name':<28} | {'Effort':<8} | {'Risk':<8} | {'Cumulative %'}")
print("-" * 75)

for i, step in enumerate(analysis['roadmap'], 1):
    category = "✅ QUICK WIN" if step['effort'] == "LOW" else "🛠️ PROJECT"
    print(f"{i:<5} | {step['name']:<28} | {step['effort']:<8} | {step['risk']:<8} | {step['cumulative_savings_pct']:>5.1f}%")

print("-" * 75)
print("💡 STRATEGIC ADVICE:")
print(f"1. Start with '{analysis['roadmap'][0]['name']}' and '{analysis['roadmap'][1]['name']}'.")
print("   These are Low Effort, Low Risk actions that secure immediate ROI.")
print(f"2. Use 'Spot Instances' only after implementing checkpointing to mitigate the HIGH risk of interruption.")
```

```python
# Cell 31: Challenge Exercise - Cost Optimization Strategy Design
print("=" * 60)
print("CHALLENGE EXERCISE 8.5.5: Design Cost Optimization Strategy")
print("=" * 60)

# 0. Định nghĩa dữ liệu đầu vào (Đảm bảo biến này tồn tại)
challenge_scenario = {
    "project": "Large Language Model Fine-tuning",
    "requirements": {
        "training_duration": 200,  # hours baseline
        "gpu_type": "A100",
        "gpu_count": 8,
        "precision": "FP32",
        "instance_type": "on-demand",
        "budget": 5000,  # USD
    },
    "constraints": {
        "max_preemption_risk": "MEDIUM",
        "min_accuracy": 0.95,
        "deadline": "2 weeks",
    }
}

# Giả định giá thuê A100 là $2.48/giờ (On-demand)
A100_PRICE = 2.48

def design_optimization_strategy(scenario):
    # 1. Tính toán chi phí Baseline
    req = scenario['requirements']
    baseline_cost = req['training_duration'] * A100_PRICE * req['gpu_count']
    
    print(f"💰 BASELINE COST: ${baseline_cost:,.2f}")
    
    # 2. Phân tích số lượng GPU tối ưu
    scaling = {1: 1.0, 2: 1.9, 4: 3.6, 8: 6.2, 16: 10.5}
    base_time_1_gpu = req['training_duration'] * scaling[req['gpu_count']]
    
    best_config = 8
    min_cost = baseline_cost
    
    for count in [4, 8, 16]:
        time_needed = base_time_1_gpu / scaling[count]
        cost = time_needed * A100_PRICE * count
        if time_needed <= 336: # 2 weeks deadline
            if cost < min_cost:
                min_cost = cost
                best_config = count

    # 3. Lựa chọn chiến lược tối ưu
    strategies = [
        {"name": "Mixed Precision (BF16/FP16)", "savings": 0.30},
        {"name": "Spot Instances", "savings": 0.60},
        {"name": "Early Stopping", "savings": 0.15}
    ]
    
    # 4. Tính toán chi phí sau tối ưu (Tích lũy)
    optimized_cost = baseline_cost
    for s in strategies:
        optimized_cost *= (1 - s['savings'])
        
    # 5. Dự báo với khoảng tin cậy 95%
    uncertainty = 0.20 
    margin = optimized_cost * uncertainty * 1.96 
    worst_case = optimized_cost + margin
    
    # 6. Kiểm tra ngân sách (SỬA LỖI KEYERROR: Lấy budget từ 'requirements')
    target_budget = scenario['requirements']['budget']
    is_under_budget = worst_case <= target_budget
    
    return {
        "baseline_cost": baseline_cost,
        "optimized_cost": optimized_cost,
        "worst_case": worst_case,
        "is_under_budget": is_under_budget,
        "best_gpu_count": best_config,
        "total_savings_pct": (1 - (optimized_cost/baseline_cost)) * 100
    }

# Chạy chiến lược
strategy = design_optimization_strategy(challenge_scenario)

print(f"\n📊 STRATEGY SUMMARY:")
print(f"   Optimal GPU Configuration: {strategy['best_gpu_count']}x A100")
print(f"   Estimated Optimized Cost:  ${strategy['optimized_cost']:,.2f}")
print(f"   Worst-Case Scenario:       ${strategy['worst_case']:,.2f}")
print(f"   Total Savings Potential:   {strategy['total_savings_pct']:.1f}%")

print(f"\n🛡️ BUDGET VERIFICATION:")
if strategy['is_under_budget']:
    print(f"   ✅ SUCCESS: Strategy meets budget of ${challenge_scenario['requirements']['budget']}")
else:
    print(f"   ❌ WARNING: Even with optimization, worst-case exceeds budget!")

print("\n💡 JUSTIFICATION:")
print("   1. Dịch chuyển sang Mixed Precision (BF16) giúp tận dụng Tensor Cores của A100, giảm 30% thời gian.")
print("   2. Sử dụng Spot Instances giúp tiết kiệm chi phí lớn nhất (60%), rủi ro Medium được quản lý bằng Checkpointing.")
print("   3. Early Stopping giúp tránh lãng phí GPU khi mô hình đã đạt độ chính xác 95% theo yêu cầu.")
```
